```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Parameters for taking entries
r_min, r_max = 100, 1500
z_min, z_max = 100, 1000
num_points = 10
variance_std = 10
```

```python
r_points = np.logspace(np.log10(r_min), np.log10(r_max), num=num_points)
z_points = np.logspace(np.log10(z_min), np.log10(z_max), num=num_points)
r, z = np.meshgrid(r_points, z_points)
r += np.random.normal(0, variance_std, size=r.shape)
z += np.random.normal(0, variance_std, size=z.shape)
plt.scatter(r, z)
plt.xlabel("Radius")
plt.ylabel("Half-Height")
plt.xlim(0, r_max*1.1)
plt.ylim(0, z_max*1.1)
plt.locator_params(axis='both', nbins=11)
plt.show()
```
