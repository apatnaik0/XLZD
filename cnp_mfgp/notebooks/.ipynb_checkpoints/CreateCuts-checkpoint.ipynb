```python
import pandas as pd
import numpy as np
import os 
from pathlib import Path

def find_repo_root(start: Path | None = None) -> Path:
    start = (start or Path.cwd()).resolve()
    for candidate in [start, *start.parents]:
        if (candidate / "PROJECT_EXPERIMENT_GUIDE.md").exists() and (candidate / "README.md").exists():
            return candidate
    raise RuntimeError("Could not find the XLZD repo root from the current working directory.")


REPO_ROOT = find_repo_root()
```

```python
# Set what files you want to cut down
filename = Path("R500_H500_2447keVgamma.csv")
data_path = REPO_ROOT / "data"
file = data_path / filename
```

```python
# Set some parameters
params = {
    "radius": 500,
    "height": 500,
    "energy_max": 3,
    "energy_min": 0.1,
}
```

```python
# Cut down the file
data = pd.read_csv(file)
data = data.copy()
len_before = len(data)

# Check if the dataset is zero on detector center or zero on detector bottom. If center, make bottom (thats what we assume on our data)
if min(data["z"]) < -1*params["height"]:
    print(f"Minimum z of {min(data["z"])}, adding {params["height"]} to z and sz")
    data["z"] = data["z"] + params["height"]
    data["sz"] = data["sz"] + params["height"]

# Calculate radius and cut based on radius and height
data.drop(data[np.sqrt(data["x"]**2 + data["y"]**2) > params["radius"]].index, inplace=True)
data.drop(data[data["z"] > 2*params["height"]].index, inplace=True)
data.drop(data[data["z"] < 0].index, inplace=True)
#data.drop(data[data["ETPC"] > params["energy_max"]].index, inplace=True)
data.drop(data[data["ETPC"] < params["energy_min"]].index, inplace=True)

# Print amount dropped
len_after = len(data)
print(f"Dropped {len_before - len_after} out of {len_before} events. {len_after} events left.")

# Save the new dataframe
new_filename = filename.with_stem(f"{filename.stem}_withcuts")
new_filepath = data_path / new_filename
if os.path.exists(new_filepath):
    os.remove(new_filepath)
data.to_csv(new_filepath, index=False)
print(f"File saved to {new_filepath}")
```
